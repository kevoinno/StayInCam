```python
import sagemaker
import boto3
from botocore.exceptions import NoCredentialsError
```

```python
# Initialize S3 client
s3 = boto3.client('s3')

# Define file and bucket details
file_name = 'face_eye_mouth_data.zip'
bucket_name = 'stayincam'
s3_file_name = 'datasets/yolo_v8_dataset.zip' # S3 key where the file will be stored
```

```python
# Upload data to S3 bucket
try:
    s3.upload_file(file_name, bucket_name, s3_file_name)
    print(f"File {file_name} uploaded to {bucket_name}/{s3_file_name}")
except FileNotFoundError:
    print("The file was not found")
except NoCredentialsError:
    print("Credentials not available")
```
